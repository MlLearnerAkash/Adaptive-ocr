```python
import os
from trdg.generators import GeneratorFromStrings

import os

def generate_images_from_words(input_file, output_dir, results_file, count, fonts):
    """
    Generate images from words listed in a file, saving images and recording results.
    
    Parameters:
      input_file (str): Path to the text file containing words (one per line).
      output_dir (str): Directory where the generated images will be saved.
      results_file (str): File path to save the image paths and corresponding labels.
      count (int): Number of images to generate per word (or overall, as supported by the generator).
      fonts (list): List of font file paths to be used by the generator.
    """
    # Create the output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Read words from the input file
    with open(input_file, 'r', encoding='utf-8') as f:
        words = [line.strip() for line in f if line.strip()]
    
    # Create the generator. Adjust the GeneratorFromStrings parameters as needed.
    generator = GeneratorFromStrings(words, count=count, fonts=fonts)
    # Open the results file for writing output information
    with open(results_file, 'w', encoding='utf-8') as rf:
        image_index = 0
        for img, label in generator:
            
            image_index += 1
            # Create a filename for the image (e.g., img_0001.png)
            filename = f"img_{image_index:04d}.png"
            image_path = os.path.join(os.getcwd(), output_dir, filename)
            # Save the generated image
            img.save(image_path)
            # Write the image path and label (separated by a tab) to the results file
            rf.write(f"{image_path}\t{label}\n")
            print(f"Saved {image_path} with label: {label}")


```
